
## 5. Customer Spending Analysis

Examine customer spending using summary statistics and a histogram.


```python
# finding Customer Spending Analysis
df["Total_Spend"].describe()
```

```python
# importing the matplotlib  and generating the histogram which shows the total spend vs no of customers
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 5))
plt.hist(df["Total_Spend"], bins=30, edgecolor="black")

plt.title("Customer Spending Distribution")
plt.xlabel("Total Spend")
plt.ylabel("Number of Customers")

plt.show()
```


## 6. Customer Rating Analysis

Analyze average ratings and visualize the number of customers for each rating.



```python
# finding the avg rating of customers
df["Average_Rating"].mean()
```

```python
# here the counting the customers and sort them by numbers availabe from 1 to 5
df["Average_Rating"].value_counts().sort_index()
```

```python
# By using the matplotlib, showing the bargraph based on avg rating vs number of customers
import matplotlib.pyplot as plt

rating_counts = df["Average_Rating"].value_counts().sort_index()

rating_counts.plot(kind="bar", figsize=(8, 5))

plt.title("Customer Rating Distribution")
plt.xlabel("Average Rating")
plt.ylabel("Number of Customers")
plt.xticks(rotation=0)
plt.show()
```


## 7. Business Analysis

Compare spending, ratings, discount usage, membership types, and satisfaction levels to identify customer behavior patterns.


```python
# finding the Membership Type vs Average Spending
df.groupby("Membership_Type")["Total_Spend"].mean().round(2)
```

```python
# finding the Satisfaction Level vs Average Rating
df.groupby("Satisfaction_Level")["Average_Rating"].mean().round(2)
```

```python
# finding the Discount Applied vs Average Spending
df.groupby("Discount_Applied")["Total_Spend"].agg(
    ["count", "mean", "median"]
).round(2)
```

```python
# finding the Membership Type vs Customer Satisfaction
pd.crosstab(
    df["Membership_Type"],
    df["Satisfaction_Level"],
    normalize="index"
).mul(100).round(2)
```

```python
# finding the Customer Spending vs Satisfaction
df.groupby("Satisfaction_Level")["Total_Spend"].agg(
    ["count", "mean", "median"]
).round(2)
```


## 8. Key Findings

- Gold members have the highest average spending among the three membership types.
- Satisfied customers have a higher average rating than neutral and unsatisfied customers.
- Average spending is very similar for customers with and without discounts.
- Gold members have the highest percentage of satisfied customers.
- Satisfied customers have higher average spending than unsatisfied customers.
